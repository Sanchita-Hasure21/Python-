```python
# WAP for generating fibonacci series 
#0,1,1,2,3,5,......
def fibonacci():
    num = int(input("Enter a number : "))
    a = 0 
    b = 1 
    for i in range (0, num+1):
        print(a, end=" ")
        a,b = b,a+b 
fibonacci()    

        
```

```python
#WAP to calculate factorial of a given number 
#4! = 4*3*2*1

def fact():
    n = int(input("Enter a number : "))
    fact = 1 
    for i in range (1, n+1):
        fact = fact * i 
    print(f'the factorial of n is {fact}')
fact()
        
        
        
        

```

```python
# WAP to calculate how many vowels are their in a string 

def count_vowel():
    s = input("enter a string : ")
    count = 0 
    vowel = 'aeiouAEIOU'
    for i in s :
        if i in vowel:
            count = count +1 
    print ("No of vowels ",count)
count_vowel()
        
    
```
